### BUBBLE SORT

```python
import time

def bubble_sort(arr):
    n = len(arr)
    a = arr.copy()

    # Step-by-step: Loop through the array n times
    for i in range(n):
        # Last i elements are already in place
        for j in range(0, n - i - 1):
            # Swap if the element found is greater than the next element
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a

# --- User Input & Execution ---
print("--- Bubble Sort Runner ---")
user_input = input("Enter numbers separated by spaces (e.g., 64 34 25 12): ")

try:
    numbers = [int(x) for x in user_input.split()]
    if not numbers:
        print("Please enter a valid list of numbers.")
    else:
        print(f"\nOriginal List: {numbers}")

        # Track execution time
        start_time = time.perf_counter()
        result = bubble_sort(numbers)
        execution_time = time.perf_counter() - start_time

        print(f"Sorted List:   {result}")
        print(f"Execution Time: {execution_time:.8f} seconds")
        print("\nTime Complexity:")
        print("  - Best Case:    O(n) (with optimization if already sorted)")
        print("  - Average Case: O(n²)")
        print("  - Worst Case:   O(n²)")
except ValueError:
    print("Error: Please enter integers only.")
```

### SELECTION SORT

```python
import time

def selection_sort(arr):
    n = len(arr)
    a = arr.copy()

    # Step-by-step: Move the boundary of the unsorted subarray
    for i in range(n):
        # Find the minimum element in remaining unsorted array
        min_idx = i
        for j in range(i + 1, n):
            if a[j] < a[min_idx]:
                min_idx = j

        # Swap the found minimum element with the first element
        a[i], a[min_idx] = a[min_idx], a[i]
    return a

# --- User Input & Execution ---
print("--- Selection Sort Runner ---")
user_input = input("Enter numbers separated by spaces (e.g., 64 34 25 12): ")

try:
    numbers = [int(x) for x in user_input.split()]
    if not numbers:
        print("Please enter a valid list of numbers.")
    else:
        print(f"\nOriginal List: {numbers}")

        # Track execution time
        start_time = time.perf_counter()
        result = selection_sort(numbers)
        execution_time = time.perf_counter() - start_time

        print(f"Sorted List:   {result}")
        print(f"Execution Time: {execution_time:.8f} seconds")
        print("\nTime Complexity:")
        print("  - Best Case:    O(n²)")
        print("  - Average Case: O(n²)")
        print("  - Worst Case:   O(n²)")
except ValueError:
    print("Error: Please enter integers only.")
```

### MERGE SORT

```python
import time

def merge_sort(arr):
    # Base case: A list of 0 or 1 elements is already sorted
    if len(arr) <= 1:
        return arr

    # Step 1: Divide the list into two halves
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])

    # Step 2: Merge the sorted halves back together
    return merge(left_half, right_half)

def merge(left, right):
    sorted_arr = []
    i = j = 0

    # Compare elements from both halves and build the sorted list
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_arr.append(left[i])
            i += 1
        else:
            sorted_arr.append(right[j])
            j += 1

    # Append any remaining elements left over from either half
    sorted_arr.extend(left[i:])
    sorted_arr.extend(right[j:])

    return sorted_arr

# --- User Input & Execution ---
print("--- Merge Sort Runner ---")
user_input = input("Enter numbers separated by spaces (e.g., 64 34 25 12): ")

try:
    numbers = [int(x) for x in user_input.split()]
    if not numbers:
        print("Please enter a valid list of numbers.")
    else:
        print(f"\nOriginal List: {numbers}")

        # Track execution time
        start_time = time.perf_counter()
        result = merge_sort(numbers)
        execution_time = time.perf_counter() - start_time

        print(f"Sorted List:   {result}")
        print(f"Execution Time: {execution_time:.8f} seconds")
        print("\nTime Complexity:")
        print("  - Best Case:    O(n log n)")
        print("  - Average Case: O(n log n)")
        print("  - Worst Case:   O(n log n)")
except ValueError:
    print("Error: Please enter integers only.")
```

### QUICK SORT

```python
import time

def quick_sort(arr):
    # Base case: arrays with 0 or 1 element are already sorted
    if len(arr) <= 1:
        return arr

    # Choosing the middle element as the pivot
    pivot = arr[len(arr) // 2]

    # Partitioning the array into three parts
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    # Recursively sort the left and right parts, then combine them
    return quick_sort(left) + middle + quick_sort(right)

# --- User Input & Execution ---
print("--- Quick Sort Runner ---")
user_input = input("Enter numbers separated by spaces (e.g., 64 34 25 12): ")

try:
    numbers = [int(x) for x in user_input.split()]
    if not numbers:
        print("Please enter a valid list of numbers.")
    else:
        print(f"\nOriginal List: {numbers}")

        # Track execution time
        start_time = time.perf_counter()
        result = quick_sort(numbers)
        execution_time = time.perf_counter() - start_time

        print(f"Sorted List:   {result}")
        print(f"Execution Time: {execution_time:.8f} seconds")
        print("\nTime Complexity:")
        print("  - Best Case:    O(n log n)")
        print("  - Average Case: O(n log n)")
        print("  - Worst Case:   O(n²) (Occurs if the pivot repeatedly choices the smallest/largest element)")
except ValueError:
    print("Error: Please enter integers only.")
```
