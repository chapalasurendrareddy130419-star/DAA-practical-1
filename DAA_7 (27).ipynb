```python
import time

def coin_change(coins, amount):
    # dp[i] = minimum number of coins needed for amount i
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    for coin in coins:
        for value in range(coin, amount + 1):
            dp[value] = min(dp[value], dp[value - coin] + 1)

    return dp[amount] if dp[amount] != float('inf') else -1


# Input
coins = list(map(int, input("Enter coin values: ").split()))
amount = int(input("Enter amount: "))

# Start time
start = time.perf_counter()

# Find minimum coins
result = coin_change(coins, amount)

# End time
end = time.perf_counter()

# Output
if result == -1:
    print("\nChange cannot be made.")
else:
    print("\nMinimum coins needed:", result)

print("Execution Time:", end - start, "seconds")
print("Time Complexity: O(n × amount)")
print("Space Complexity: O(amount)")
```
