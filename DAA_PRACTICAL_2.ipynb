# BINARY SEARCH

```python
import time

def binary_search(arr, target):
    """
    Performs binary search on a sorted list to find a target value.

    Args:
        arr (list): A sorted list of numbers.
        target (int): The number to search for.

    Returns:
        int: The index of the target if found, otherwise -1.
    """
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1
```

### Complexity Analysis

*   **Time Complexity:** O(log n)
    *   In each step, binary search divides the search space in half. This logarithmic behavior makes it very fast for large datasets.
    *   Imagine searching for a word in a dictionary: you don't check every page; you open roughly to the middle, then decide if you need to go forward or backward. Binary search does the same with numbers!

*   **Space Complexity:** O(1)
    *   Binary search only uses a few variables (like `low`, `high`, `mid`), regardless of how large the input list `n` is. This means it requires a constant amount of extra memory.

```python
# Get user input for the sorted list
while True:
    try:
        list_str = input("Enter a sorted list of numbers separated by spaces (e.g., 10 20 30 40): ")
        user_list = sorted([int(x) for x in list_str.split()])
        break
    except ValueError:
        print("Invalid input. Please enter numbers only.")

# Get user input for the target value
while True:
    try:
        target_value = int(input("Enter the number you want to search for: "))
        break
    except ValueError:
        print("Invalid input. Please enter an integer.")

print(f"\nSearching for {target_value} in the list: {user_list}")

# Measure execution time
start_time = time.perf_counter()
index = binary_search(user_list, target_value)
end_time = time.perf_counter()

execution_time = (end_time - start_time) * 1000 # Convert to milliseconds

# Display results
if index != -1:
    print(f"Target {target_value} found at index {index}.")
else:
    print(f"Target {target_value} not found in the list.")

print(f"Execution Time: {execution_time:.6f} milliseconds")
```

# LINEAR SEARCH

```python
# Linear Search in Python
# Program to search an element in a list using Linear Search

import time

# Function for Linear Search
def linear_search(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return -1

# Taking input from the user
n = int(input("Enter the number of elements: "))

arr = []
print("Enter the elements:")
for i in range(n):
    arr.append(int(input()))

key = int(input("Enter the element to search: "))

# Start measuring execution time
start_time = time.perf_counter()

# Perform Linear Search
result = linear_search(arr, key)

# End measuring execution time
end_time = time.perf_counter()

# Display the result
if result != -1:
    print(f"\nElement {key} found at index {result}.")
else:
    print(f"\nElement {key} was not found in the list.")

# Display execution time
execution_time = end_time - start_time
print(f"Execution Time: {execution_time:.10f} seconds")

# Complexity
print("\nTime Complexity:")
print("Best Case    : O(1)")
print("Average Case : O(n)")
print("Worst Case   : O(n)")

print("\nSpace Complexity:")
print("O(1) - Constant space")
```
